### Extracting Numeric Columns

```python
numeric_columns = df.select_dtypes(include=['number']).columns
print(f"Numeric Columns : {numeric_columns}")
```

### Displaying Stastics for Each Categorical Column

```python
for i in range(len(categorical_columns)):
    print(f"Statistical Summary for {categorical_columns[i]}\n-------------------------------------------------------------------------------------")
    print(df[[categorical_columns[i]] + list(numeric_columns)].groupby(categorical_columns[i]).describe())
```

## Statistical Description for Categorical Variables

### Grouping by Species

```python
grouped = df_iris.groupby('Species')
```

### Function to calculate statistics for particular species

```python
def getStats(x):
  return pd.Series({
      '---' : '------------------------------------------------',
      'count' : x.count(),
      'mean' : x.mean(),
      'median' : x.median(),
      'std' : x.std(),
      'min' : x.min(),
      'max' : x.max(),
      'percentile_25': x.quantile(0.25),
      'percentile_50': x.quantile(0.5),
      'percentile_75': x.quantile(0.75)
  })
```

### Displaying Statistical Details for each Species

```python
for i in ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']:
  print(f"Statistical Summary for {i}\n_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_-_\n")
  print(grouped[i].apply(getStats))
```
