# Import Required Libraries

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

# Load the Titanic Dataset

```python
titanic = sns.load_dataset('titanic')
```

# Plot Histogram for Ticket Fare

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=titanic, x='fare', kde=True, bins=30, color='skyblue')

plt.title('Distribution of Titanic Ticket Fares')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')

plt.show()
```

# Fare Distribution by Passenger Class

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=titanic, x='pclass', y='fare', palette='pastel')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

# Fare Distribution by Survival Status

```python
# Fare vs Survived
plt.figure(figsize=(10, 6))
sns.violinplot(data=titanic, x='survived', y='fare', palette='Set2')

plt.title('Fare Distribution by Survival Status')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Fare')
plt.show()
```

# Correlation Heatmap

```python
# Correlation matrix
plt.figure(figsize=(10, 6))
sns.heatmap(titanic.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")

plt.title('Correlation Heatmap of Numeric Features')
plt.show()
```
