# Data Prepocessing

## Identifying NULL Values

```python
display(df.isnull())
```

```python
print(f"No. of NULL Values per Column\n{df.isnull().sum()}\n-------------------------------\nTotal No. of NULL Values : {df.isnull().sum().sum()}")
```

## Imputing NULL Values

```python
df['age'] = df['age'].fillna(df['age'].median())
```

```python
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])
df['embarked_town'] = df['embark_town'].fillna(df['embark_town'].mode()[0])
```

```python
df.drop(columns=['deck'], inplace=True)
```

```python
print(f"No. of NULL Values per Column\n{df.isnull().sum()}\n-------------------------------\nTotal No. of NULL Values : {df.isnull().sum().sum()}")
```

## Detecting Outliers

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='age', data=df, color='skyblue')
plt.title('Boxplot of Age (Outlier Detection)')
plt.xlabel('Age')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Converting to Suitable Data Types

```python
display(df.dtypes)
```

```python
print("Memory Usage Before Converting to Category Columns")
df.memory_usage(deep=True).sum()
```

```python
display(df.nunique())
```

```python
cat_cols = df.columns[df.nunique() <= 5] #Only selecting features with unique values less than 5
df[cat_cols] = df[cat_cols].astype('category')
```

```python
display(df.dtypes)
```

```python
print("Memory Usage After Converting to Category Columns")
df.memory_usage(deep=True).sum()
```

# Quick EDA

## KDE Plot

```python
plt.figure(figsize=(12,8))
sns.kdeplot(data=df, x='age', fill=True, hue='survived')
plt.show()
```

```python
survived_count = df[df['survived'] == 0].groupby('sex').size()
print(f"No. of People who did not Survive:\n{survived_count}")
```

## Histogram of Ticket Fare Distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['fare'], bins=10, kde=True, color='skyblue')
plt.title('Fare Distribution of Titanic Passengers')
plt.xlabel('Fare')
plt.ylabel('Count')
plt.show()
```

## CountPlot : Survival Count by Gender

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='sex', hue='survived', data=df)
plt.title('Survival Count by Gender')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

## Boxplot: Fare by Passenger Class

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='pclass', y='fare', data=df, hue='pclass')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

## Violin Plot: Age Distribution by Survival

```python
plt.figure(figsize=(8, 5))
sns.violinplot(x='survived', y='age', data=df, palette='Set2')
plt.title('Age Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()
```

## Countplot: Survival by Embarkation Port

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='embarked', hue='survived', data=df)
plt.title('Survival by Embarkation Port')
plt.xlabel('Port of Embarkation')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

## Survival Rate by Passenger Class and Gender (Stacked Analysis)

```python
plt.figure(figsize=(8, 5))
sns.catplot(data=df, x='pclass', hue='sex', col='survived', kind='count', height=5, aspect=1)
plt.suptitle('Survival Count by Class and Gender', y=1.05)
plt.show()
```

## Heatmap of Correlation Between Numerical Features

```python
plt.figure(figsize=(8, 6))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```
