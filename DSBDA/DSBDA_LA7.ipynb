# Install & Import Dependencies

```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
```

## Downloading Required nltk Datasets

```python
# Download necessary NLTK datasets
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
```

# Input Text & Lowercasing

```python
# Input Text
text = "It is a truth universally acknowledged, that a single man in possession of a good fortune, must be in want of a wife."

# Convert to lowercase
text = text.lower()
print("Lowercased Text:\n", text)
```

# Remove Punctuation

```python
# Remove Punctuation
text_p = "".join([char for char in text if char not in string.punctuation])
print("Text without Punctuation:\n", text_p)
```

# Tokenization

```python
words = word_tokenize(text_p)
print("Tokens:\n", words)
```

# Stopword Removal

```python
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]
print("Filtered Words (Stopwords Removed):\n", filtered_words)
```

# Stemming

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("Stemmed Words:\n", stemmed_words)
```

# Lemmatization

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
print("Lemmatized Words:\n", lemmatized_words)
```

# Part of Speech Tagging

```python
pos_tags = pos_tag(filtered_words)
print("POS Tags:\n", pos_tags)
```

#Term Frequency and Inverse Document Frequency

## Term Frequency

```python
from collections import Counter
import math

term_counts = Counter(words) #I've stored tokens in variable name word
total_terms = len(words)

tf = {term: count / total_terms for term, count in term_counts.items()}
```

## Inverse Document Frequency

```python
N = 1
idf = {term: math.log((N + 1) / (1 + 1)) + 1 for term in term_counts}  # All = 1.0

# Step 3: TF-IDF
tf_idf = {term: tf[term] * idf[term] for term in tf}

# Output
for term, value in tf_idf.items():
    print(f"{term}: {value:.4f}")
```
